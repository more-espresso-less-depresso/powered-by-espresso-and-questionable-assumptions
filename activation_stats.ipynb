```python
model_path = ""
dataset_path = ""
general_result_folder = f"results/{model_path}/{dataset_path}"
```

```python
result_folder = f"{general_result_folder}/grouped"
```

```python
safe_file = f"{result_folder}/h_safe_activations.pt"
unsafe_file = f"{result_folder}/h_unsafe_activations.pt"
refusal_file = f"{result_folder}/h_refusal_activations.pt"

if not os.path.exists(safe_file) or not os.path.exists(unsafe_file) or not os.path.exists(refusal_file):
    print("Activations not found. Please run the activation extraction script first.")
else:
    safe_activations = load_pt_to_stack(safe_file)
    safe_activations = safe_activations.float().detach().numpy()
    unsafe_activations = load_pt_to_stack(unsafe_file)
    unsafe_activations = unsafe_activations.float().detach().numpy()
    refusal_activations = load_pt_to_stack(refusal_file)
    refusal_activations = refusal_activations.float().detach().numpy()

    visualize_triple_activations(safe_activations, unsafe_activations, refusal_activations, "Responses", len(safe_activations), f"{general_result_folder}/responses")
```

```python
result_folder = f"{general_result_folder}/all"
```

```python
all_b_file = f"{result_folder}/b_activations.pt"
all_h_file = f"{result_folder}/h_activations.pt"

if not os.path.exists(all_b_file) or not os.path.exists(all_h_file):
    print("Activations not found. Please run the activation extraction script first.")
else:
    b_activations = load_pt_to_stack(all_b_file)
    b_activations = b_activations.float().detach().numpy()
    h_activations = load_pt_to_stack(all_h_file)
    h_activations = h_activations.float().detach().numpy()

    visualize_comparison_activations(b_activations, h_activations, "Prompts", len(b_activations), f"{general_result_folder}/prompts")
```
